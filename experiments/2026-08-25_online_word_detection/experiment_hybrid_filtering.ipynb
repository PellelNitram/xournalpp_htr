# Hybrid Classifier-Filtered Spatial Clustering for Word Stroke Detection

**Approach**: Use a Gradient Boosting classifier to pre-filter unlikely stroke pairs, then run agglomerative clustering on bbox distances only among pairs the classifier deems plausible. This combines the robustness of spatial clustering with the classifier's ability to reject clearly wrong pairs.

**Two hyperparameters** (tuned jointly via grid search on training pages per CV fold):
- `prob_threshold`: minimum P(same word) from the classifier to keep a pair
- `dist_threshold`: agglomerative clustering distance threshold on the filtered distance matrix

**Evaluation**: Leave-one-page-out cross-validation (6 folds).

## 1. Data Loading

```python
import gzip
import json
import warnings
from dataclasses import dataclass
from itertools import combinations
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from bs4 import BeautifulSoup as bs
from bs4 import XMLParsedAsHTMLWarning
from huggingface_hub import snapshot_download
from scipy.spatial.distance import cdist
from sklearn.cluster import AgglomerativeClustering
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import StandardScaler
from tabulate import tabulate

warnings.filterwarnings("ignore", category=XMLParsedAsHTMLWarning)

_TEXT_CLASSES = {"word", "digit", "mathematical_expression"}


@dataclass
class Stroke:
    x: np.ndarray
    y: np.ndarray


@dataclass
class PageData:
    page_id: str
    strokes: list[Stroke]
    word_groups: list[list[int]]


def _parse_strokes_per_layer(path: Path) -> list[list[list[Stroke]]]:
    with gzip.open(path, "r") as f:
        content = f.read().decode("utf-8")
    soup = bs(content, "lxml")
    pages = []
    for page in soup.find_all("page"):
        layers = []
        for layer in page.find_all("layer"):
            layer_strokes = []
            for stroke_el in layer.find_all("stroke"):
                coords = np.fromstring(stroke_el.text, sep=" ").reshape(-1, 2).T
                layer_strokes.append(Stroke(x=coords[0], y=coords[1]))
            layers.append(layer_strokes)
        pages.append(layers)
    return pages


def load_benchmark_pages() -> list[PageData]:
    local_dir = Path(
        snapshot_download("PellelNitram/xournalpp_htr_benchmark", repo_type="dataset")
    )
    data_dir = local_dir / "data"
    xopp_files = sorted(data_dir.glob("*.xopp")) + sorted(data_dir.glob("*.xoj"))

    pages = []
    for xopp_path in xopp_files:
        gt_path = xopp_path.with_suffix("").with_suffix(".gt.json")
        if not gt_path.exists():
            continue
        strokes_per_layer = _parse_strokes_per_layer(xopp_path)
        with open(gt_path) as f:
            gt = json.load(f)

        page_annotations: dict[int, list[list[int]]] = {}
        for ann in gt["annotations"]:
            if ann["class"] not in _TEXT_CLASSES:
                continue
            page_idx = ann["page_index"]
            layer_idx = ann["layer_index"]
            offset = sum(
                len(strokes_per_layer[page_idx][li]) for li in range(layer_idx)
            )
            global_indices = [offset + si for si in ann["stroke_indices"]]
            page_annotations.setdefault(page_idx, []).append(global_indices)

        for page_idx, layer_list in enumerate(strokes_per_layer):
            flat_strokes = [s for layer in layer_list for s in layer]
            if not flat_strokes:
                continue
            word_groups = page_annotations.get(page_idx, [])
            page_id = f"{xopp_path.stem}_p{page_idx}"
            pages.append(
                PageData(page_id=page_id, strokes=flat_strokes, word_groups=word_groups)
            )

    return pages


pages = load_benchmark_pages()
for p in pages:
    n_word_strokes = len({si for wg in p.word_groups for si in wg})
    print(
        f"{p.page_id}: {len(p.strokes)} strokes, {len(p.word_groups)} words, {n_word_strokes} word strokes"
    )
```

## 2. Stroke & Pair Features

```python
@dataclass
class StrokeFeatures:
    bbox: tuple[float, float, float, float]
    centroid: tuple[float, float]
    n_points: int
    path_length: float
    bbox_width: float
    bbox_height: float
    bbox_area: float
    aspect_ratio: float


def compute_stroke_features(stroke: Stroke) -> StrokeFeatures:
    min_x, max_x = float(stroke.x.min()), float(stroke.x.max())
    min_y, max_y = float(stroke.y.min()), float(stroke.y.max())
    cx, cy = float(stroke.x.mean()), float(stroke.y.mean())
    n = len(stroke.x)
    diffs = np.sqrt(np.diff(stroke.x) ** 2 + np.diff(stroke.y) ** 2)
    path_length = float(diffs.sum())
    w = max_x - min_x
    h = max_y - min_y
    area = w * h
    aspect = w / h if h > 1e-6 else w / 1e-6
    return StrokeFeatures(
        bbox=(min_x, min_y, max_x, max_y),
        centroid=(cx, cy),
        n_points=n,
        path_length=path_length,
        bbox_width=w,
        bbox_height=h,
        bbox_area=area,
        aspect_ratio=aspect,
    )


def stroke_bbox(stroke: Stroke) -> tuple[float, float, float, float]:
    return (
        float(stroke.x.min()),
        float(stroke.y.min()),
        float(stroke.x.max()),
        float(stroke.y.max()),
    )


def bbox_min_edge_distance(bbox_a, bbox_b) -> float:
    ax0, ay0, ax1, ay1 = bbox_a
    bx0, by0, bx1, by1 = bbox_b
    dx = max(0.0, max(ax0 - bx1, bx0 - ax1))
    dy = max(0.0, max(ay0 - by1, by0 - ay1))
    return np.sqrt(dx * dx + dy * dy)


def compute_pairwise_bbox_distances(strokes: list[Stroke]) -> np.ndarray:
    n = len(strokes)
    bboxes = [stroke_bbox(s) for s in strokes]
    dist = np.zeros((n, n))
    for i in range(n):
        for j in range(i + 1, n):
            d = bbox_min_edge_distance(bboxes[i], bboxes[j])
            dist[i, j] = d
            dist[j, i] = d
    return dist


FEATURE_NAMES = [
    "min_point_dist",
    "bbox_edge_dist",
    "centroid_dist",
    "horizontal_gap",
    "vertical_gap",
    "horizontal_overlap_ratio",
    "vertical_overlap_ratio",
    "centroid_angle",
    "stroke_index_diff",
    "area_ratio",
    "aspect_ratio_diff",
    "path_length_ratio",
    "n_points_ratio",
]


def compute_pair_features(
    stroke_a: Stroke,
    feat_a: StrokeFeatures,
    idx_a: int,
    stroke_b: Stroke,
    feat_b: StrokeFeatures,
    idx_b: int,
) -> np.ndarray:
    pts_a = np.column_stack([stroke_a.x, stroke_a.y])
    pts_b = np.column_stack([stroke_b.x, stroke_b.y])
    min_point_dist = cdist(pts_a, pts_b).min()

    ax0, ay0, ax1, ay1 = feat_a.bbox
    bx0, by0, bx1, by1 = feat_b.bbox
    dx = max(0.0, max(ax0 - bx1, bx0 - ax1))
    dy = max(0.0, max(ay0 - by1, by0 - ay1))
    bbox_edge_dist = np.sqrt(dx * dx + dy * dy)

    cx_diff = feat_a.centroid[0] - feat_b.centroid[0]
    cy_diff = feat_a.centroid[1] - feat_b.centroid[1]
    centroid_dist = np.sqrt(cx_diff**2 + cy_diff**2)

    x_overlap = max(0.0, min(ax1, bx1) - max(ax0, bx0))
    min_width = min(feat_a.bbox_width, feat_b.bbox_width) + 1e-6
    horizontal_overlap_ratio = x_overlap / min_width

    y_overlap = max(0.0, min(ay1, by1) - max(ay0, by0))
    min_height = min(feat_a.bbox_height, feat_b.bbox_height) + 1e-6
    vertical_overlap_ratio = y_overlap / min_height

    centroid_angle = abs(np.arctan2(cy_diff, cx_diff + 1e-9))
    stroke_index_diff = abs(idx_a - idx_b)

    area_a = feat_a.bbox_area + 1e-6
    area_b = feat_b.bbox_area + 1e-6
    area_ratio = max(area_a, area_b) / min(area_a, area_b)
    aspect_ratio_diff = abs(feat_a.aspect_ratio - feat_b.aspect_ratio)

    pl_a = feat_a.path_length + 1e-6
    pl_b = feat_b.path_length + 1e-6
    path_length_ratio = max(pl_a, pl_b) / min(pl_a, pl_b)

    n_a, n_b = feat_a.n_points, feat_b.n_points
    n_points_ratio = max(n_a, n_b) / max(min(n_a, n_b), 1)

    return np.array(
        [
            min_point_dist,
            bbox_edge_dist,
            centroid_dist,
            dx,
            dy,
            horizontal_overlap_ratio,
            vertical_overlap_ratio,
            centroid_angle,
            stroke_index_diff,
            area_ratio,
            aspect_ratio_diff,
            path_length_ratio,
            n_points_ratio,
        ]
    )
```

## 3. Build Training Data

```python
def build_page_data(
    page: PageData,
) -> tuple[np.ndarray, np.ndarray, list[tuple[int, int]]]:
    stroke_feats = [compute_stroke_features(s) for s in page.strokes]
    stroke_to_word = {}
    for wi, wg in enumerate(page.word_groups):
        for si in wg:
            stroke_to_word[si] = wi

    n = len(page.strokes)
    pairs = list(combinations(range(n), 2))

    X = np.zeros((len(pairs), len(FEATURE_NAMES)))
    y = np.zeros(len(pairs), dtype=int)

    for k, (i, j) in enumerate(pairs):
        X[k] = compute_pair_features(
            page.strokes[i],
            stroke_feats[i],
            i,
            page.strokes[j],
            stroke_feats[j],
            j,
        )
        wi = stroke_to_word.get(i, -1)
        wj = stroke_to_word.get(j, -1)
        y[k] = 1 if (wi == wj and wi != -1) else 0

    return X, y, pairs


print("Building pairwise data per page...")
page_data = {}
for page in pages:
    X, y, pairs = build_page_data(page)
    page_data[page.page_id] = (X, y, pairs)
    n_pos = y.sum()
    n_neg = len(y) - n_pos
    print(f"  {page.page_id}: {len(pairs)} pairs, {n_pos} same-word, {n_neg} diff-word")
```

## 4. Evaluation Metrics

```python
@dataclass
class GroupingMetrics:
    precision: float
    recall: float
    f1: float
    stroke_accuracy: float
    n_gt_words: int
    n_pred_clusters: int


def evaluate_grouping(
    predicted_clusters: list[set[int]],
    gt_word_groups: list[list[int]],
    n_strokes: int,
    jaccard_threshold: float = 0.5,
) -> GroupingMetrics:
    gt_sets = [set(wg) for wg in gt_word_groups]

    candidates = []
    for i, pred in enumerate(predicted_clusters):
        for j, gt in enumerate(gt_sets):
            intersection = len(pred & gt)
            union = len(pred | gt)
            if union > 0:
                jaccard = intersection / union
                if jaccard >= jaccard_threshold:
                    candidates.append((jaccard, i, j))

    candidates.sort(key=lambda x: x[0], reverse=True)

    matched_pred, matched_gt = set(), set()
    for _, pi, gi in candidates:
        if pi not in matched_pred and gi not in matched_gt:
            matched_pred.add(pi)
            matched_gt.add(gi)

    n_matched = len(matched_pred)
    n_pred = len(predicted_clusters)
    n_gt = len(gt_sets)

    precision = n_matched / n_pred if n_pred > 0 else 0.0
    recall = n_matched / n_gt if n_gt > 0 else 0.0
    f1 = (
        2 * precision * recall / (precision + recall)
        if (precision + recall) > 0
        else 0.0
    )

    stroke_to_gt = {}
    for gi, gt in enumerate(gt_sets):
        for si in gt:
            stroke_to_gt[si] = gi
    stroke_to_pred = {}
    for pi, pred in enumerate(predicted_clusters):
        for si in pred:
            stroke_to_pred[si] = pi

    correct = 0
    total_word_strokes = len(stroke_to_gt)
    for si, gi in stroke_to_gt.items():
        if si in stroke_to_pred:
            pi = stroke_to_pred[si]
            pred_set = predicted_clusters[pi]
            gt_set = gt_sets[gi]
            intersection = len(pred_set & gt_set)
            union = len(pred_set | gt_set)
            if union > 0 and intersection / union >= jaccard_threshold:
                correct += 1

    stroke_accuracy = correct / total_word_strokes if total_word_strokes > 0 else 0.0

    return GroupingMetrics(
        precision=precision,
        recall=recall,
        f1=f1,
        stroke_accuracy=stroke_accuracy,
        n_gt_words=n_gt,
        n_pred_clusters=n_pred,
    )
```

## 5. Hybrid Classifier-Filtered Clustering

1. Train GB classifier on training pages
2. Predict P(same word) for all pairs on a page
3. Build filtered distance matrix: bbox distance where P >= prob_threshold, else infinity
4. Run agglomerative clustering (single linkage) with dist_threshold
5. Grid-search both thresholds on training pages

```python
def build_filtered_distance_matrix(
    strokes: list[Stroke],
    pairs: list[tuple[int, int]],
    proba: np.ndarray,
    prob_threshold: float,
) -> np.ndarray:
    """Build a distance matrix where filtered-out pairs have infinite distance."""
    n = len(strokes)
    bbox_dist = compute_pairwise_bbox_distances(strokes)

    pair_to_idx = {(i, j): k for k, (i, j) in enumerate(pairs)}

    filtered_dist = np.full((n, n), 1e6)
    np.fill_diagonal(filtered_dist, 0.0)

    for (i, j), k in pair_to_idx.items():
        if proba[k] >= prob_threshold:
            filtered_dist[i, j] = bbox_dist[i, j]
            filtered_dist[j, i] = bbox_dist[j, i]

    return filtered_dist


def cluster_with_filtered_matrix(
    strokes: list[Stroke],
    pairs: list[tuple[int, int]],
    proba: np.ndarray,
    prob_threshold: float,
    dist_threshold: float,
) -> list[set[int]]:
    """Cluster strokes using filtered distance matrix + agglomerative clustering."""
    if len(strokes) == 0:
        return []
    if len(strokes) == 1:
        return [{0}]

    filtered_dist = build_filtered_distance_matrix(
        strokes, pairs, proba, prob_threshold
    )

    clustering = AgglomerativeClustering(
        n_clusters=None,
        metric="precomputed",
        linkage="single",
        distance_threshold=dist_threshold,
    )
    labels = clustering.fit_predict(filtered_dist)

    clusters: dict[int, set[int]] = {}
    for i, label in enumerate(labels):
        clusters.setdefault(label, set()).add(i)

    return list(clusters.values())


def tune_hybrid_thresholds(
    clf,
    scaler,
    train_pages: list[PageData],
    page_data: dict,
    prob_candidates=None,
    dist_candidates=None,
) -> tuple[float, float]:
    """Grid search over prob_threshold and dist_threshold on training pages."""
    if prob_candidates is None:
        prob_candidates = np.arange(0.1, 0.9, 0.1)
    if dist_candidates is None:
        dist_candidates = np.arange(1.0, 50.0, 2.0)

    best_prob_t = 0.5
    best_dist_t = 5.0
    best_f1 = -1.0

    for prob_t in prob_candidates:
        for dist_t in dist_candidates:
            f1s = []
            for page in train_pages:
                X_val, _, pairs_val = page_data[page.page_id]
                X_val_scaled = scaler.transform(X_val)
                proba = clf.predict_proba(X_val_scaled)[:, 1]
                clusters = cluster_with_filtered_matrix(
                    page.strokes, pairs_val, proba, prob_t, dist_t
                )
                metrics = evaluate_grouping(
                    clusters, page.word_groups, len(page.strokes)
                )
                f1s.append(metrics.f1)
            mean_f1 = np.mean(f1s)
            if mean_f1 > best_f1:
                best_f1 = mean_f1
                best_prob_t = prob_t
                best_dist_t = dist_t

    return float(best_prob_t), float(best_dist_t)
```

## 6. LOPO Cross-Validation

```python
results = []

for i, test_page in enumerate(pages):
    train_pages = pages[:i] + pages[i + 1 :]

    X_train = np.vstack([page_data[p.page_id][0] for p in train_pages])
    y_train = np.concatenate([page_data[p.page_id][1] for p in train_pages])

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)

    clf = GradientBoostingClassifier(
        n_estimators=200, max_depth=4, learning_rate=0.1, random_state=42
    )
    clf.fit(X_train_scaled, y_train)

    prob_t, dist_t = tune_hybrid_thresholds(clf, scaler, train_pages, page_data)

    X_test, _, pairs_test = page_data[test_page.page_id]
    X_test_scaled = scaler.transform(X_test)
    proba = clf.predict_proba(X_test_scaled)[:, 1]

    clusters = cluster_with_filtered_matrix(
        test_page.strokes, pairs_test, proba, prob_t, dist_t
    )
    metrics = evaluate_grouping(clusters, test_page.word_groups, len(test_page.strokes))

    results.append((test_page.page_id, prob_t, dist_t, metrics, clusters))
    print(
        f"Fold {i+1}/6: {test_page.page_id} — prob_t={prob_t:.2f}, dist_t={dist_t:.1f}, F1={metrics.f1:.3f}"
    )
```

## 7. Results

```python
headers = [
    "Page",
    "Prob Thresh",
    "Dist Thresh",
    "Precision",
    "Recall",
    "F1",
    "Stroke Acc",
    "GT Words",
    "Pred Clusters",
]
rows = []
for page_id, prob_t, dist_t, m, _ in results:
    rows.append(
        [
            page_id,
            f"{prob_t:.2f}",
            f"{dist_t:.1f}",
            f"{m.precision:.3f}",
            f"{m.recall:.3f}",
            f"{m.f1:.3f}",
            f"{m.stroke_accuracy:.3f}",
            m.n_gt_words,
            m.n_pred_clusters,
        ]
    )

precisions = [m.precision for _, _, _, m, _ in results]
recalls = [m.recall for _, _, _, m, _ in results]
f1s = [m.f1 for _, _, _, m, _ in results]
stroke_accs = [m.stroke_accuracy for _, _, _, m, _ in results]

rows.append(
    [
        "MEAN",
        "",
        "",
        f"{np.mean(precisions):.3f} ± {np.std(precisions):.3f}",
        f"{np.mean(recalls):.3f} ± {np.std(recalls):.3f}",
        f"{np.mean(f1s):.3f} ± {np.std(f1s):.3f}",
        f"{np.mean(stroke_accs):.3f} ± {np.std(stroke_accs):.3f}",
        "",
        "",
    ]
)

print(tabulate(rows, headers=headers, tablefmt="grid"))

print("\n")
print("=" * 70)
print("COMPARISON SUMMARY")
print("=" * 70)


def summarize(name, f1s, precisions, recalls, stroke_accs):
    return [
        name,
        f"{np.mean(precisions):.3f} ± {np.std(precisions):.3f}",
        f"{np.mean(recalls):.3f} ± {np.std(recalls):.3f}",
        f"{np.mean(f1s):.3f} ± {np.std(f1s):.3f}",
        f"{np.mean(stroke_accs):.3f} ± {np.std(stroke_accs):.3f}",
    ]


comparison = [
    [
        "Spatial Baseline",
        "0.762 ± 0.175",
        "0.734 ± 0.213",
        "0.733 ± 0.178",
        "0.641 ± 0.259",
    ],
    [
        "GB + Conn. Components",
        "0.738 ± 0.219",
        "0.741 ± 0.318",
        "0.719 ± 0.288",
        "0.768 ± 0.286",
    ],
    summarize("Hybrid (GB + Spatial)", f1s, precisions, recalls, stroke_accs),
]

print(
    tabulate(
        comparison,
        headers=["Method", "Precision", "Recall", "F1", "Stroke Acc"],
        tablefmt="grid",
    )
)
```

## 8. Visualize Predictions vs GT

```python
def plot_comparison(page: PageData, clusters: list[set[int]], title: str = ""):
    fig, (ax_gt, ax_pred) = plt.subplots(1, 2, figsize=(16, 8))
    cmap = plt.colormaps["tab20"]

    ax_gt.set_title("Ground Truth")
    word_stroke_set = {si for wg in page.word_groups for si in wg}
    for gi, wg in enumerate(page.word_groups):
        color = cmap(gi % 20)
        for si in wg:
            s = page.strokes[si]
            ax_gt.plot(s.x, -s.y, color=color, linewidth=1.5)
    for si, s in enumerate(page.strokes):
        if si not in word_stroke_set:
            ax_gt.plot(s.x, -s.y, color="lightgray", linewidth=0.5)
    ax_gt.set_aspect("equal")
    ax_gt.axis("off")

    ax_pred.set_title("Predicted Clusters")
    clustered_strokes = {si for c in clusters for si in c}
    for ci, cluster in enumerate(clusters):
        color = cmap(ci % 20)
        for si in cluster:
            s = page.strokes[si]
            ax_pred.plot(s.x, -s.y, color=color, linewidth=1.5)
    for si, s in enumerate(page.strokes):
        if si not in clustered_strokes:
            ax_pred.plot(s.x, -s.y, color="lightgray", linewidth=0.5)
    ax_pred.set_aspect("equal")
    ax_pred.axis("off")

    if title:
        fig.suptitle(title, fontsize=14)
    plt.tight_layout()
    plt.show()


for page_id, prob_t, dist_t, m, clusters in results:
    page = next(p for p in pages if p.page_id == page_id)
    plot_comparison(
        page,
        clusters,
        f"{page_id} (prob_t={prob_t:.2f}, dist_t={dist_t:.1f}, F1={m.f1:.3f})",
    )
```

## 9. ONNX Export

Train the scaler + classifier on **all 6 pages** and export as a single ONNX model.
The feature extraction and clustering logic are pure Python/NumPy — only the learned
pairwise classifier ships as a model artifact.

The exported model takes a `(N, 13)` float32 array of pairwise features and outputs
the probability of each pair being "same word".

```python
import onnxruntime as ort
from skl2onnx import to_onnx
from sklearn.pipeline import Pipeline

X_all = np.vstack([page_data[p.page_id][0] for p in pages])
y_all = np.concatenate([page_data[p.page_id][1] for p in pages])

print(
    f"Training on all data: {X_all.shape[0]} pairs, {y_all.sum()} positive, {len(y_all) - y_all.sum()} negative"
)

pipeline = Pipeline(
    [
        ("scaler", StandardScaler()),
        (
            "clf",
            GradientBoostingClassifier(
                n_estimators=200,
                max_depth=4,
                learning_rate=0.1,
                random_state=42,
            ),
        ),
    ]
)
pipeline.fit(X_all, y_all)

train_proba = pipeline.predict_proba(X_all)[:, 1]
train_pred = (train_proba >= 0.5).astype(int)
train_acc = (train_pred == y_all).mean()
print(f"Train accuracy: {train_acc:.4f}")

onnx_model = to_onnx(
    pipeline, X_all[:1].astype(np.float32), options={id(pipeline): {"zipmap": False}}
)

export_path = Path("hybrid_word_detector.onnx")
with open(export_path, "wb") as f:
    f.write(onnx_model.SerializeToString())
print(
    f"Exported ONNX model to {export_path} ({export_path.stat().st_size / 1024:.1f} KB)"
)

# Verify: compare sklearn vs ONNX predictions on a sample
sess = ort.InferenceSession(str(export_path))
sample = X_all[:100].astype(np.float32)

sklearn_proba = pipeline.predict_proba(sample)
onnx_out = sess.run(None, {"X": sample})
onnx_proba = onnx_out[1]  # probabilities

max_diff = np.abs(sklearn_proba - onnx_proba).max()
print(f"Max sklearn vs ONNX probability difference: {max_diff:.2e}")
assert max_diff < 1e-5, f"ONNX export mismatch: max diff = {max_diff}"
print("ONNX verification passed!")
```
